```polyglot-notebook
public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers();
}

public class Employee
{
    private string name;
    private string position;
    private double salary;

    public string Name
    {
        get {return name;}
        set
        {

        if (!string.IsNullOrWhiteSpace(value))
            name = value;
        else
            throw new ArgumentException ("Введите имя");
        }
    }

    public string Position
    {
        get {return position;}
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                position = value;
            else
                throw new ArgumentException("Требуется должность");
        }
    }
    public double Salary
    {
        get {return salary;}
        set
        {
            if (value>=0)
                salary = value;
            else
                throw new ArgumentOutOfRangeException("Зарплата не отрицательная");
        }
    }

    public static int TotalEmployees {get; private set;} = 0;

    public Employee(string name, string position, double salery)
    {
        Name = name;
        Position = position;
        Salary = salary;
        TotalEmployees++; 
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name}, Должность: {Position}, Зарплата: {Salary:F2} Руб.");
    }

    public static void ShowTotal()
    {
        Console.WriteLine($"Всего сотрудников: {TotalEmployees}");
    }

    public class Manager  : Employee
    {
        public string Department{get; set;}

        public Manager(string name, string position, double salary, string department)
            : base(name, position, salary)
        {
            Department = department;
        }

        public override void PrintInfo()
        {
            base.PrintInfo();
            Console.WriteLine($"Отдел: {Department}");
        }
    }

    public class ProjectManager : Manager, IReportable, ITeamLeader
    {
        public string ProjectName {get; set;}

        public ProjectManager(string name, string position, double salary, string departament, string projectName)
            : base(name, position, salary, departament)
            {
                ProjectName = projectName;
            }
        public override void PrintInfo()
        {
            base.PrintInfo();
            Console.WriteLine($"Проект: {ProjectName}");
        }

        public string GenerateReport()
        {
            return $"Отчёт по проекту '{ProjectName}'   Менеджер: {Name}   Отдел: {Department}";
        }

        public string[] GetTeamMembers()
        {
            return new string[] { "Иванов", "Петрова", "Сидоров"};
        }
    }
}

Employee employee = new Employee("Алексей", "Разработчик", 80000);
Manager manager = new Manager("Мария", "Менеджер", 120000, "IT");
ProjectManager pjm = new ProjectManager("Иван", "Проектный менеджер", 150000, "Разработка", "Сайт");

employee.PrintInfo();
Console.WriteLine();

manager.PrintInfo();
Console.WriteLine();

pjm.PrintInfo();
Console.WriteLine();

string report = pmj.GenerateReport();
string[] team = pmj.GetTeamMembers();

Console.WriteLine(report);
foreach (var member in team)
{
    Console.WriteLine($"Участник команды: {member}");
}

Console.WriteLine();
Employee.ShowTotal();
```
