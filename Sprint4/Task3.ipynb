```
public interface IAnimal
{
    void WakeUp();
    void Sleep();
    int GetEnergyLevel();
}

public abstract class Animal : IAnimal
{
    private string name;
    private int age;
    private string habitat;

    public string Name
    {
        get { return name; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                name = value;
            else
                throw new ArgumentException("Нужно имя");
        }
    }

    public int Age
    {
        get { return age; }
        set
        {
            if (value >= 0)
                age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не отрицательный");
        }
    }

    public string Habitat
    {
        get { return habitat; }
        set { habitat = value; }
    }

    protected int energyLevel;

    public Animal(string name, int age, string habitat)
    {
        Name = name;
        Age = age;
        Habitat = habitat;
        energyLevel = 100;
    }

    public abstract void WakeUp();
    public abstract void Sleep();

    public virtual int GetEnergyLevel()
    {
        return energyLevel;
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string habitat, string breed) : base(name, age, habitat)
    {
        Breed = breed;
    }

    public override void WakeUp()
    {
        energyLevel = 100;
        Console.WriteLine($"Собака {Name} проснулась");
    }

    public override void Sleep()
    {
        energyLevel = 0;
        Console.WriteLine($"Собака {Name} уснула");
    }

    public void Run(int distance)
    {
        energyLevel -= 15;
        Console.WriteLine($"Собака {Name} пробежала {distance} метров");
    }
}

public class Cat : Animal
{
    public string Color { get; set; }

    public Cat(string name, int age, string habitat, string color) : base(name, age, habitat)
    {
        Color = color;
    }

    public override void WakeUp()
    {
        energyLevel = 100;
        Console.WriteLine($"Кошка {Name} проснулась");
    }

    public override void Sleep()
    {
        energyLevel = 0;
        Console.WriteLine($"Кошка {Name} уснула");
    }

    public void Purr()
    {
        energyLevel -= 5;
        Console.WriteLine($"Кoшка {Name} мурчит");
    }
}

public class Bird : Animal
{
    public string WingSpan { get; set; }

    public Bird(string name, int age, string habitat, string wingSpan) : base(name, age, habitat)
    {
        WingSpan = wingSpan;
    }

    public override void WakeUp()
    {
        energyLevel = 100;
        Console.WriteLine($"Птица {Name} проснулась");
    }

    public override void Sleep()
    {
        energyLevel = 0;
        Console.WriteLine($"Птица {Name} уснула");
    }

    public void Fly(int height)
    {
        energyLevel -= 20;
        Console.WriteLine($"Птица {Name} взлетела на высоту {height} метров");
    }
}

public class AnimalService
{
    private readonly IAnimal animal;

    public AnimalService(IAnimal animal)
    {
        this.animal = animal;
    }

    public void UseAnimal()
    {
        animal.WakeUp();
        Console.WriteLine($"Уровень энергии: {animal.GetEnergyLevel()}");
        animal.Sleep();
    }
}

Dog laika = new Dog("Лайка", 5, "Дом", "Овчарка");
Cat murka = new Cat("Мурка", 3, "Дом", "Белый");
Bird kesha = new Bird("Кеша", 2, "Лес", "50 см");

AnimalService laikaService = new AnimalService(laika);
laikaService.UseAnimal();
laika.Run(50);

Console.WriteLine();

AnimalService murkaService = new AnimalService(murka);
murkaService.UseAnimal();
murka.Purr();

Console.WriteLine();

AnimalService keshaService = new AnimalService(kesha);
keshaService.UseAnimal();
kesha.Fly(30);
```
