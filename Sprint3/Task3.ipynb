```
public class Animal
{
    private string name;
    private int age;
    private string habitat;

    public string Name
    {
        get { return name; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                name = value;
            else
                throw new ArgumentException("Нужно имя");
        }
    }

    public int Age
    {
        get { return age; }
        set
        {
            if (value >= 0)
                age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не отрицательный");
        }
    }

    public string Habitat
    {
        get { return habitat; }
        set { habitat = value; }
    }

    public Animal(string name, int age, string habitat)
    {
        Name = name;
        Age = age;
        Habitat = habitat;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издаёт звук");
    }

    public void MakeSound(int count)
    {
        Console.WriteLine($"{Name} издаёт звук {count} раз");
    }

    public void MakeSound(string mood)
    {
        Console.WriteLine($"{Name} издаёт звук с настроением {mood}");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Вoзраст: {Age}, Среда: {Habitat}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string habitat, string breed) : base(name, age, habitat)
    {
        Breed = breed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} Лает");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
}

public class Cat : Animal
{
    public string Color { get; set; }

    public Cat(string name, int age, string habitat, string color) : base(name, age, habitat)
    {
        Color = color;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} Мяукает");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Окрас: {Color}");
    }
}

public class Bird : Animal
{
    public string WingSpan { get; set; }

    public Bird(string name, int age, string habitat, string wingSpan) : base(name, age, habitat)
    {
        WingSpan = wingSpan;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} Чирикает");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {WingSpan}");
    }
}

public class AnimalCollection<T> where T : Animal
{
    private List<T> animals = new List<T>();

    public void Add(T animal)
    {
        animals.Add(animal);
    }

    public void Remove(T animal)
    {
        animals.Remove(animal);
    }

    public void DisplayAnimals()
    {
        foreach (var animal in animals)
        {
            animal.DisplayInfo();
            animal.MakeSound();
            Console.WriteLine();
        }
    }
}

Dog laika = new Dog("Лайка", 5, "Дом", "Овчарка");
Cat murka = new Cat("Мурка", 3, "Дом", "Белый");
Bird kesha = new Bird("Кеша", 2, "Лес", "50 см");

AnimalCollection<Animal> myAnimals = new AnimalCollection<Animal>();
myAnimals.Add(laika);
myAnimals.Add(murka);
myAnimals.Add(kesha);

myAnimals.DisplayAnimals();

Animal genericAnimal = new Animal("Зверь", 1, "Лес");
genericAnimal.MakeSound();
genericAnimal.MakeSound(3);
genericAnimal.MakeSound("Радость");
```
