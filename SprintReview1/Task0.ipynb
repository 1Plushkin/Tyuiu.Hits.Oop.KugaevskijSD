```polyglot-notebook
public class Product
{
    private string _name;
    private decimal _price;
    private int _quantity;
    protected string _category;
    internal string _description;

    public string Name
    {
        get { return _name; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Нужно название");
        }
    }

    public decimal Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
                _price = value;
            else
                throw new ArgumentOutOfRangeException("Цена не отрицательная");
        }
    }

    public int Quantity
    {
        get { return _quantity; }
        set
        {
            if (value >= 0)
                _quantity = value;
            else
                throw new ArgumentOutOfRangeException("Количество не отрицательное");
        }
    }

    public string Category
    {
        get { return _category; }
        set { _category = value; }
    }

    public string Description
    {
        get { return _description; }
        set { _description = value; }
    }

    public Product()
    {
        Name = "Неизвестно";
        Price = 0;
        Quantity = 0;
        Category = "Общее";
        Description = "Без описания";
    }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice() => Price * Quantity;

    public void UpdateQuantity(int amount)
    {
        Quantity += amount;
        Console.WriteLine($"Количество товара '{Name}' обновлено: {Quantity} шт.");
    }

    public string GetProductInfo() =>
        $"Название: {Name}, Цена: {Price:F2} Руб., Количество: {Quantity} шт., Категория: {Category}, Описание: {Description}";
}

public class Seller
{
    private string _name;
    private string _employeeId;
    protected decimal _salary;
    internal string _contactInfo;

    public string Name
    {
        get { return _name; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Введите имя");
        }
    }

    public string EmployeeId
    {
        get { return _employeeId; }
        set { _employeeId = value; }
    }

    public decimal Salary
    {
        get { return _salary; }
        set
        {
            if (value >= 0)
                _salary = value;
            else
                throw new ArgumentOutOfRangeException("Зарплата не отрицательная");
        }
    }

    public string ContactInfo
    {
        get { return _contactInfo; }
        set { _contactInfo = value; }
    }

    public List<Product> Products { get; set; } = new List<Product>();

    public Seller()
    {
        Name = "Неизвестно";
        EmployeeId = "000";
        Salary = 0;
        ContactInfo = "Не указано";
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        Products.Add(product);
        Console.WriteLine($"Товар '{product.Name}' добавлен продавцу {Name}");
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(-quantity);
            Console.WriteLine($"{Name} продал {quantity} шт. товара '{product.Name}'");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара '{product.Name}' на складе");
        }
    }

    public string GetSellerInfo() =>
        $"Продавец: {Name}, ID: {EmployeeId}, Зарплата: {Salary:F2} Руб., Контакт: {ContactInfo}";
}

public class Store
{
    private string _storeName;
    private string _location;
    public string StoreHours { get; set; }

    public string StoreName
    {
        get { return _storeName; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _storeName = value;
            else
                throw new ArgumentException("Название магазина");
        }
    }

    public string Location
    {
        get { return _location; }
        set { _location = value; }
    }

    public List<Seller> Sellers { get; set; } = new List<Seller>();

    public Store()
    {
        StoreName = "Неизвестно";
        Location = "Не указано";
        StoreHours = "09:00 - 21:00";
    }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        StoreHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        Sellers.Add(seller);
        Console.WriteLine($"Продавец '{seller.Name}' принят на работу в магазин '{StoreName}'");
    }

    public void ListProducts()
    {
        Console.WriteLine($"\n Товары магазина '{StoreName}'");
        foreach (var seller in Sellers)
        {
            foreach (var product in seller.Products)
            {
                Console.WriteLine(product.GetProductInfo());
            }
        }
    }

    public string GetStoreInfo() =>
        $"Магазин: {StoreName}, Адрес: {Location}, Часы работы: {StoreHours}";
}

Product milk = new Product("Молоко", 65, 300, "Молочное", "Молоко");
Seller semen = new Seller("Семён", "1", 35000, "semen@mail.ru");
Store store = new Store("Лента", "ул. Октябрьская, 25", "Круглосуточно");

semen.AddProduct(milk);
store.AddSeller(semen);
Console.WriteLine();
Console.WriteLine(store.GetStoreInfo());
Console.WriteLine(semen.GetSellerInfo());
Console.WriteLine();
store.ListProducts();
Console.WriteLine();
semen.SellProduct(milk, 73);
store.ListProducts();
```
