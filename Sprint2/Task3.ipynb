```
using System;

public interface IAnimal
{
    string Name { get; set; }
    int Age { get; set; }
    void DisplayInfo();
}

public interface IPet
{
    string OwnerName { get; set; }
    void ShowOwner();
}

public interface IFlyable
{
    string WingSpan { get; set; }
    void Fly();
}

public interface ISwimmable
{
    int MaxDepth { get; set; }
    void Swim();
}

public abstract class Animal
{
    private string name;
    private int age;
    private string habitat;
    protected string category;

    public string Name
    {
        get { return name; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                name = value;
            else
                throw new ArgumentException("Нужно имя");
        }
    }

    public int Age
    {
        get { return age; }
        set
        {
            if (value >= 0)
                age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не отрицательный");
        }
    }

    public string Habitat
    {
        get { return habitat; }
        set { habitat = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public static int TotalAnimals { get; private set; } = 0;

    protected Animal(string name, int age, string habitat, string category)
    {
        Name = name;
        Age = age;
        Habitat = habitat;
        Category = category;
        TotalAnimals++;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет, Среда: {Habitat}, Категория: {Category}");
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издаёт звук");
    }

    public virtual void Eat(string food)
    {
        Console.WriteLine($"{Name} ест {food}");
    }

    public static void ShowTotal()
    {
        Console.WriteLine($"Всего животных: {TotalAnimals}");
    }
}

public class Dog : Animal, IAnimal, IPet
{
    public string Breed { get; set; }
    public string OwnerName { get; set; }

    public Dog(string name, int age, string habitat, string category, string breed, string ownerName)
        : base(name, age, habitat, category)
    {
        Breed = breed;
        OwnerName = ownerName;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} Лает");
    }

    public void ShowOwner()
    {
        Console.WriteLine($"Хозяин {Name}: {OwnerName}");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}, Хозяин: {OwnerName}");
    }
}

public class Cat : Animal, IAnimal, IPet
{
    public string Color { get; set; }
    public string OwnerName { get; set; }

    public Cat(string name, int age, string habitat, string category, string color, string ownerName)
        : base(name, age, habitat, category)
    {
        Color = color;
        OwnerName = ownerName;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} Мяукает");
    }

    public void ShowOwner()
    {
        Console.WriteLine($"Хозяин {Name}: {OwnerName}");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Окрас: {Color}, Хозяин: {OwnerName}");
    }
}

public class Bird : Animal, IAnimal, IFlyable
{
    public string WingSpan { get; set; }

    public Bird(string name, int age, string habitat, string category, string wingSpan)
        : base(name, age, habitat, category)
    {
        WingSpan = wingSpan;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} Чирикает");
    }

    public void Fly()
    {
        Console.WriteLine($"{Name} летает, размах крыльев: {WingSpan}");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {WingSpan}");
    }
}

public class Parrot : Bird, IAnimal, IFlyable, IPet
{
    public bool CanTalk { get; set; }
    public string OwnerName { get; set; }

    public Parrot(string name, int age, string habitat, string category, string wingSpan, bool canTalk, string ownerName)
        : base(name, age, habitat, category, wingSpan)
    {
        CanTalk = canTalk;
        OwnerName = ownerName;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} Говорит");
    }

    public void ShowOwner()
    {
        Console.WriteLine($"Хозяин {Name}: {OwnerName}");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Умеет говорить: {CanTalk}, Хозяин: {OwnerName}");
    }
}

public class Penguin : Animal, IAnimal, ISwimmable
{
    public int MaxDepth { get; set; }

    public Penguin(string name, int age, string habitat, string category, int maxDepth)
        : base(name, age, habitat, category)
    {
        MaxDepth = maxDepth;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} Кричит");
    }

    public void Swim()
    {
        Console.WriteLine($"{Name} плаваеt на глубине до {MaxDepth} м");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Макс. глубина: {MaxDepth} м");
    }
}

Dog laika = new Dog("Лайка", 5, "Дом", "Млекопитающее", "Овчарка", "Иван");
Cat murka = new Cat("Мурка", 3, "Дом", "Млекопитающее", "Белый", "Мария");
Bird kesha = new Bird("Кеша", 2, "Лес", "Птица", "50 см");
Parrot gosha = new Parrot("Гоша", 4, "Дом", "Птица", "40 см", true, "Пётр");
Penguin pingu = new Penguin("Пингу", 6, "Антарктида", "Птица", 200);

laika.DisplayInfo();
laika.MakeSound();
laika.ShowOwner();

murka.DisplayInfo();
murka.MakeSound();
murka.ShowOwner();

kesha.DisplayInfo();
kesha.MakeSound();
kesha.Fly();

gosha.DisplayInfo();
gosha.MakeSound();
gosha.Fly();
gosha.ShowOwner();

pingu.DisplayInfo();
pingu.MakeSound();
pingu.Swim();

Animal.ShowTotal();
```
