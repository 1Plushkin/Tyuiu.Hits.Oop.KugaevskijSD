```polyglot-notebook
using System;
public class Plant
{
    public string Name 
    {
        get { return Name; }
        set { Name = value; }
    }    
    public string Type 
     {
        get { return Type; }
        set { Type = value; }
    }    
    public float Height
     {
        get { return Height; }
        set { Height = value; }
    }    


    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветет.");
    }

    public void Grow(float amount)
    {
        Height += amount;
        Console.WriteLine($"Растение {Name} выросло на {amount} см. Текущая высота: {Height} см.");
    }
}

public class Flower : Plant
{
    public string Color 
      {
        get { return Color; }
        set { Color = value; }
    }    
    public int PetalCount 
      {
        get { return PetalCount; }
        set { PetalCount = value; }
    }    

    public Flower(string name, float height, string color, int petalCount) 
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать.");
    }
}

public class Tree : Plant
{
    public int Age 
    {
        get { return Age; }
        set { Age = value; }
    }    
    public string LeafType 
    {
        get { return LeafType; }
        set { LeafType = value; }
    }    

    public Tree(string name, float height, int age, string leafType) 
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowOlder(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} стало старше на {years} лет. Новый возраст: {Age} лет.");
    }
}


Flower rose = new Flower("Роза", 15.5f, "Красный", 20);
Tree oak = new Tree("Дуб", 200.0f, 5, "Лиственные");

Console.WriteLine("Демонстрация роста ");
rose.Grow(5.0f);
oak.Grow(10.0f);
oak.GrowOlder(1);

Console.WriteLine("\nДемонстрация увядания");
rose.Wither();

Console.WriteLine("\nДемонстрация полиморфизма");

Plant[] garden = { rose, oak };

foreach (Plant plant in garden)
{
    plant.Bloom();
}
```
