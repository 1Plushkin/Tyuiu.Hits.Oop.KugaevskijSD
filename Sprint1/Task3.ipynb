```
public class Item
{
    private int _itemId;
    private string _name;
    private decimal _price;

    public int ItemId
    {
        get { return _itemId; }
        set
        {
            if (value > 0)
                _itemId = value;
            else
                throw new ArgumentOutOfRangeException("ID должен быть больше нуля");
        }
    }

    public string Name
    {
        get { return _name; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Название не может быть пустым");
        }
    }

    public decimal Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
                _price = value;
            else
                throw new ArgumentOutOfRangeException("Цена не отрицательная");
        }
    }

    public Item() 
    {
        ItemId = 1;
        Name = "Неизвестно";
        Price = 0;
    }

    public Item(int id, string name, decimal price)
    {
        ItemId = id;
        Name = name;
        Price = price;
    }

    public virtual string GetDetails() =>
        $"ID: {ItemId}, Название: {Name}, Цена: {Price:F2} Руб.";

    public virtual decimal CalculateDiscount() => Price * 0.05m;

    public virtual void ApplyDiscount(decimal discount)
    {
        Price -= Math.Min(discount, Price);
        Console.WriteLine($"Скидка {discount:F2} Руб. Новая цена: {Price:F2} Руб.");
    }
}

public class SingleItem : Item
{
    private string _unitMeasure;

    public string UnitMeasure
    {
        get { return _unitMeasure; }
        set { _unitMeasure = value; }
    }

    public SingleItem() : base()
    {
        UnitMeasure = "шт.";
    }

    public SingleItem(int id, string name, decimal price, string unit)
        : base(id, name, price)
    {
        UnitMeasure = unit;
    }

    public override string GetDetails() =>
        $"{base.GetDetails()}, Ед. изм.: {UnitMeasure}";
}

public class PackageItem : Item
{
    private int _quantityPerPackage;

    public int QuantityPerPackage
    {
        get { return _quantityPerPackage; }
        set
        {
            if (value > 0)
                _quantityPerPackage = value;
            else
                throw new ArgumentOutOfRangeException("Количество должно быть больше нуля");
        }
    }

    public PackageItem() : base()
    {
        QuantityPerPackage = 1;
    }

    public PackageItem(int id, string name, decimal price, int qty)
        : base(id, name, price)
    {
        QuantityPerPackage = qty;
    }

    public override decimal CalculateDiscount() =>
        Price * Math.Min(0.02m * QuantityPerPackage, 0.30m);

    public override string GetDetails() =>
        $"{base.GetDetails()}, Количество в пакете: {QuantityPerPackage} шт.";
}

public class SpecialItem : Item
{
    private DateTime _discountExpirationDate;

    public DateTime DiscountExpirationDate
    {
        get { return _discountExpirationDate; }
        set { _discountExpirationDate = value; }
    }

    public SpecialItem() : base()
    {
        DiscountExpirationDate = DateTime.Now.AddDays(30);
    }

    public SpecialItem(int id, string name, decimal price, DateTime expDate)
        : base(id, name, price)
    {
        DiscountExpirationDate = expDate;
    }

    public override void ApplyDiscount(decimal discount)
    {
        if (DateTime.Now > DiscountExpirationDate)
        {
            Console.WriteLine($"Скидка истекла {DiscountExpirationDate:dd.MM.yyyy}");
            return;
        }
        Console.WriteLine($"Скидка действует до {DiscountExpirationDate:dd.MM.yyyу}");
        base.ApplyDiscount(discount);
    }

    public override string GetDetails() =>
        $"{base.GetDetails()}, Скидка до: {DiscountExpirationDate:dd.MM.yyyy}";
}

Item[] items =
{
    new SingleItem(1, "Молоко", 89.90m, "Литры"),
    new PackageItem(2, "Яблоки", 250m, 6),
    new SpecialItem(3, "Смартфон", 45000m, DateTime.Now.AddDays(10))
};

foreach (var item in items)
{
    Console.WriteLine(item.GetDetails());
    var discount = item.CalculateDiscount();
    Console.WriteLine($"Расчётная скидка: {discount:F2} Руб.");
    item.ApplyDiscount(discount);
    Console.WriteLine();
}

SingleItem milk = (SingleItem)items[0];
PackageItem apples = (PackageItem)items[1];
SpecialItem phone = (SpecialItem)items[2];

milk.Price = 100m;
apples.QuantityPerPackage = 10;
phone.DiscountExpirationDate = DateTime.Now.AddDays(5);

Console.WriteLine("=== После взаимодействия ===");
Console.WriteLine(milk.GetDetails());
Console.WriteLine(apples.GetDetails());
Console.WriteLine(phone.GetDetails());
```
