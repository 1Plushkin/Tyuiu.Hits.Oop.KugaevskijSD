<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Item в C#, который будет представлять информацию о
товарах, которые могут быть заказаны или возвращены. На основе этого класса
разработать 2-3 производных класса, демонстрирующих принципы наследования и
полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и
методы, а также переопределены некоторые методы базового класса для
демонстрации полиморфизма.

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public class Item
{
    public int ItemId { get; set; }
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Item(int id, string name, decimal price)
    {
        ItemId = id;
        Name = name;
        Price = price;
    }

    public virtual string GetDetails() =>
        $"ID: {ItemId}, Название: {Name}, Цена: {Price:F2} Руб.";

    public virtual decimal CalculateDiscount() => Price * 0.05m;

    public virtual void ApplyDiscount(decimal discount)
    {
        Price -= Math.Min(discount, Price);
        Console.WriteLine($"Скидка {discount:F2} Руб. Новая цена: {Price:F2} Руб.");
    }
}

public class SingleItem : Item
{
    public string UnitMeasure { get; set; }

    public SingleItem(int id, string name, decimal price, string unit)
        : base(id, name, price) => UnitMeasure = unit;

    public override string GetDetails() =>
        $"{base.GetDetails()}, Ед. изм.: {UnitMeasure}";
}

public class PackageItem : Item
{
    public int QuantityPerPackage { get; set; }

    public PackageItem(int id, string name, decimal price, int qty)
        : base(id, name, price) => QuantityPerPackage = qty;

    public override decimal CalculateDiscount() =>
        Price * Math.Min(0.02m * QuantityPerPackage, 0.30m);

    public override string GetDetails() =>
        $"{base.GetDetails()}, Количество в пакете: {QuantityPerPackage} шт.";
}

public class SpecialItem : Item
{
    public DateTime DiscountExpirationDate { get; set; }

    public SpecialItem(int id, string name, decimal price, DateTime expDate)
        : base(id, name, price) => DiscountExpirationDate = expDate;

    public override void ApplyDiscount(decimal discount)
    {
        if (DateTime.Now > DiscountExpirationDate)
        {
            Console.WriteLine($"Скидка истекла {DiscountExpirationDate:dd.MM.yyyy}");
            return;
        }
        Console.WriteLine($"Скидка действует до {DiscountExpirationDate:dd.MM.yyyy}");
        base.ApplyDiscount(discount);
    }

    public override string GetDetails() =>
        $"{base.GetDetails()}, Скидка до: {DiscountExpirationDate:dd.MM.yyyy}";
}

Item[] items =
{
    new SingleItem(1, "Молоко", 89.90m, "Литры"),
    new PackageItem(2, "Яблоки", 250m, 6),
    new SpecialItem(3, "Смартфон", 45000m, DateTime.Now.AddDays(10))
};

foreach (var item in items)
{
    Console.WriteLine(item.GetDetails());
    var discount = item.CalculateDiscount();
    Console.WriteLine($"Расчётная скидка: {discount:F2} Руб.");
    item.ApplyDiscount(discount);
    Console.WriteLine();
}
```
