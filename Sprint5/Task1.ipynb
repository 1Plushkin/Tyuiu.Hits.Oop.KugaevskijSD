<h1 style="color:DodgerBlue">HashSet<T></h1>



`HashSet<T>` в C# — это коллекция, которая хранит уникальные элементы без определенного порядка. Она полезна, когда нужно быстро проверять наличие элемента или добавлять только уникальные элементы. Давайте рассмотрим, как с ней работать на примере классов `Transport`, `Car`, и `Bike`.

### Создание классов Transport, Car, и Bike

Сначала определим базовый класс `Transport` и его наследники `Car` и `Bike`:

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public override bool Equals(object obj)
    {
        // Простая проверка равенства базируется на модели и производителе
        if (obj is Transport transport)
        {
            return Model == transport.Model && Manufacturer == transport.Manufacturer;
        }
        return false;
    }

    public override int GetHashCode()
    {
        // Создание хеш-кода на основе модели и производителя
        return HashCode.Combine(Model, Manufacturer);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}
```

### Работа с HashSet<T>

Теперь рассмотрим, как использовать `HashSet<T>` для хранения объектов `Transport`.

```csharp
using System;
using System.Collections.Generic;


// Создаем HashSet для уникальных объектов транспортных средств
HashSet<Transport> transportSet = new HashSet<Transport>();

// Добавляем объекты Car и Bike в HashSet
var car1 = new Car("Model S", "Tesla", 4);
var car2 = new Car("Model 3", "Tesla", 4);
 var bike1 = new Bike("Mountain Explorer", "Giant", true);

transportSet.Add(car1);
transportSet.Add(car2);
transportSet.Add(bike1);

// Попробуем добавить повторяющийся объект
var duplicateCar = new Car("Model S", "Tesla", 4);
bool isAdded = transportSet.Add(duplicateCar);

Console.WriteLine($"Was duplicate car added? {isAdded}"); // false

// Перебираем и выводим информацию обо всех уникальных транспортных средствах
foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

```

### Объяснения и комментарии:

- **Уникальность объектов**: `HashSet<T>` не допускает дублирования. Если вы пытаетесь добавить элемент, который уже присутствует (определяется методом `Equals` и `GetHashCode`), он не добавляется.

- **Методы `Equals` и `GetHashCode`**: Эти методы переопределены для корректной работы с `HashSet`. `Equals` определяет, когда два объекта считаются равными, а `GetHashCode` возвращает хеш-код, используемый для быстрого поиска элемента.

- **Быстрая проверка на наличие**: Благодаря хешированию, `HashSet` позволяет быстро проверять, существует ли элемент в коллекции (`O(1)` — амортизированное время).

- **Отсутствие порядка**: Элементы в `HashSet` не имеют определенного порядка, в отличие от `List<T>`.

`HashSet<T>` идеально подходит для сценариев, где требуется исключительно уникальность элементов и быстрое выполнение операций поиска.

````
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public override bool Equals(object obj)
    {
        if (obj is Transport transport)
        {
            return Model == transport.Model && Manufacturer == transport.Manufacturer;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Manufacturer);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет багажник: {HasCarrier}");
    }
}

HashSet<Transport> transportSet = new HashSet<Transport>();

var car1 = new Car("Model S", "Tesla", 4);
var car2 = new Car("Model 3", "Tesla", 4);
var bike1 = new Bike("Mountain Eхplorer", "Giant", true);

transportSet.Add(car1);
transportSet.Add(car2);
transportSet.Add(bike1);

var duplicateCar = new Car("Model S", "Tesla", 4);
bool isAdded = transportSet.Add(duplicateCar);

Console.WriteLine($"Дубликат добавлен? {isAdded}");

Console.WriteLine("Уникальный транспорт");
foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}
````

````
public interface IAnimal
{
    string Name { get; set; }
    int Age { get; set; }
    void DisplayInfo();
}

public interface IPet
{
    string Owner { get; set; }
    void ShowOwner();
}

public interface IFlyable
{
    void Fly();
}

public interface ISwimmable
{
    void Swim();
}

public class Animal : IAnimal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Price { get; set; }

    public Animal(string name, int age, double price)
    {
        Name = name;
        Age = age;
        Price = price;
    }

    public override bool Equals(object obj)
    {
        if (obj is Animal animal)
        {
            return Name == animal.Name && Age == animal.Age && Price == animal.Price;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Age, Price);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Цена: {Price:F2} Руб");
    }

    public virtual void MakeSound(int times)
    {
        for (int i = 0; i < times; i++)
            Console.WriteLine("...");
    }

    public virtual void MakeSound(string message)
    {
        Console.WriteLine($"Животное говорит: {message}");
    }
}

public class Dog : Animal, IPet
{
    public string Breed { get; set; }
    public string Owner { get; set; }

    public Dog(string name, int age, double price, string breed, string owner)
        : base(name, age, price)
    {
        Breed = breed;
        Owner = owner;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}, Хозяин: {Owner}");
    }

    public override void MakeSound(int times)
    {
        for (int i = 0; i < times; i++)
            Console.WriteLine("Гав!");
    }

    public override void MakeSound(string message)
    {
        Console.WriteLine($"Собака {Name} лает: {message}");
    }

    public void ShowOwner()
    {
        Console.WriteLine($"Хозяин собаки {Name}: {Owner}");
    }
}

public class Cat : Animal, IPet
{
    public string Color { get; set; }
    public string Owner { get; set; }

    public Cat(string name, int age, double price, string color, string owner)
        : base(name, age, price)
    {
        Color = color;
        Owner = owner;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Окрас: {Color}, Хозяин: {Owner}");
    }

    public override void MakeSound(int times)
    {
        for (int i = 0; i < times; i++)
            Console.WriteLine("Мяу!");
    }

    public override void MakeSound(string message)
    {
        Console.WriteLine($"Кошка {Name} мяукает: {message}");
    }

    public void ShowOwner()
    {
        Console.WriteLine($"Хозяин кошки {Name}: {Owner}");
    }
}

public class Bird : Animal, IFlyable
{
    public double WingSpan { get; set; }

    public Bird(string name, int age, double price, double wingSpan)
        : base(name, age, price)
    {
        WingSpan = wingSpan;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {WingSpan:F2} см");
    }

    public override void MakeSound(int times)
    {
        for (int i = 0; i < times; i++)
            Console.WriteLine("Чирик!");
    }

    public override void MakeSound(string message)
    {
        Console.WriteLine($"Птица {Name} поёт: {message}");
    }

    public virtual void Fly()
    {
        Console.WriteLine($"{Name} летает.");
    }
}

public class Parrot : Bird
{
    public string Vocabulary { get; set; }

    public Parrot(string name, int age, double price, double wingSpan, string vocabulary)
        : base(name, age, price, wingSpan)
    {
        Vocabulary = vocabulary;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Словарный запас: {Vocabulary}");
    }

    public override void MakeSound(string message)
    {
        Console.WriteLine($"Попугай {Name} повторяет: {message}");
    }

    public override void Fly()
    {
        Console.WriteLine($"{Name} летает по комнате.");
    }
}

public class Penguin : Bird, ISwimmable
{
    public Penguin(string name, int age, double price, double wingSpan)
        : base(name, age, price, wingSpan) { }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine("Пингвин не умеет летать, но отлично плавает.");
    }

    public override void MakeSound(int times)
    {
        for (int i = 0; i < times; i++)
            Console.WriteLine("Га-га!");
    }

    public void Swim()
    {
        Console.WriteLine($"{Name} плавает.");
    }

    public override void Fly()
    {
        Console.WriteLine($"{Name} не умеет летать.");
    }
}

public class AnimalCollection<T> where T : Animal
{
    private HashSet<T> _items = new HashSet<T>();

    public bool Add(T animal)
    {
        return _items.Add(animal);
    }

    public void DisplayAll()
    {
        foreach (var animal in _items)
        {
            animal.DisplayInfo();
            Console.WriteLine();
        }
    }
}

HashSet<Animal> animalSet = new HashSet<Animal>();

var dog1 = new Dog("Лайка", 3, 25000.00, "Овчарка", "Иван");
var dog2 = new Dog("Отгадай", 2, 30000.00, "Лабрадор", "Мария");
var dog3 = new Dog("Котя", 4, 28000.00, "Хаски", "Пётр");
var cat1 = new Cat("Мурка", 2, 15000.00, "Белый", "Анна");
var cat2 = new Cat("Рыжик", 1, 12000.00, "Рыжий", "Сергей");
var cat3 = new Cat("Фелемонтий", 5, 18000.00, "Серый", "Ольга");
var parrot1 = new Parrot("Кеша", 2, 20000.00, 25.5, "Привет, как дела?");
var parrot2 = new Parrot("Гоша", 3, 22000.00, 28.0, "Кто там?");
var penguin1 = new Penguin("Пингу", 4, 35000.00, 40.0);

animalSet.Add(dog1);
animalSet.Add(dog2);
animalSet.Add(dog3);
animalSet.Add(cat1);
animalSet.Add(cat2);
animalSet.Add(cat3);
animalSet.Add(parrot1);
animalSet.Add(parrot2);
animalSet.Add(penguin1);

var duplicateDog = new Dog("Лайка", 3, 25000.00, "Овчарка", "Иван");
bool isDogAdded = animalSet.Add(duplicateDog);

Console.WriteLine($"Дубликат собаки добавлен? {isDogAdded}");

Console.WriteLine("Уникальные животные");
foreach (var animal in animalSet)
{
    animal.DisplayInfo();
    Console.WriteLine();
}

AnimalCollection<Dog> dogCollection = new AnimalCollection<Dog>();
dogCollection.Add(new Dog("Лайка", 3, 25000.00, "Овчарка", "Иван"));
dogCollection.Add(new Dog("Отгадай", 2, 30000.00, "Лабрадор", "Мария"));
dogCollection.Add(new Dog("Котя", 4, 28000.00, "Хаски", "Пётр"));
dogCollection.Add(new Dog("Лайка", 3, 25000.00, "Овчарка", "Иван"));

Console.WriteLine("Коллекция собак");
dogCollection.DisplayAll();
````
