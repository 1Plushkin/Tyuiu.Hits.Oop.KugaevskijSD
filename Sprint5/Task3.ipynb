```
public delegate void AnimalActionDelegate(string message);
public delegate void DisplayDelegate();

public interface IAnimal
{
    void WakeUp();
    void Sleep();
    int GetEnergyLevel();
}

public interface IPet : IAnimal
{
    void Play();
}

public interface IFlyable
{
    void Fly();
}

public interface ISwimmable
{
    void Swim();
}

public abstract class Animal : IAnimal
{
    private string name;
    private int age;
    private double price;
    private string habitat;

    public string Name
    {
        get { return name; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                name = value;
            else
                throw new ArgumentException("Нужно имя");
        }
    }

    public int Age
    {
        get { return age; }
        set
        {
            if (value >= 0)
                age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не отрицательный");
        }
    }

    public double Price
    {
        get { return price; }
        set
        {
            if (value >= 0)
                price = value;
            else
                throw new ArgumentOutOfRangeException("Цена не отрицательная");
        }
    }

    public string Habitat
    {
        get { return habitat; }
        set { habitat = value; }
    }

    protected int energyLevel;

    public Animal(string name, int age, double price, string habitat)
    {
        Name = name;
        Age = age;
        Price = price;
        Habitat = habitat;
        energyLevel = 100;
    }

    public event AnimalActionDelegate EnergyChanged;
    public event AnimalActionDelegate ActionPerformed;

    public abstract void WakeUp();
    public abstract void Sleep();

    public virtual int GetEnergyLevel()
    {
        return energyLevel;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Цена: {Price:F2} Руб, Среда: {Habitat}, Энергия: {energyLevel}");
    }

    protected void OnEnergyChanged(string message)
    {
        EnergyChanged?.Invoke(message);
    }

    protected void OnActionPerformed(string message)
    {
        ActionPerformed?.Invoke(message);
    }

    public override bool Equals(object obj)
    {
        if (obj is Animal animal)
        {
            return Name == animal.Name && Age == animal.Age;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Age);
    }
}

public class Dog : Animal, IPet
{
    public string Breed { get; set; }
    public string Owner { get; set; }
    public bool IsTrained { get; set; }

    public Dog(string name, int age, double price, string habitat, string breed, string owner, bool isTrained)
        : base(name, age, price, habitat)
    {
        Breed = breed;
        Owner = owner;
        IsTrained = isTrained;
    }

    void IAnimal.WakeUp()
    {
        energyLevel = 100;
        Console.WriteLine($"{Name} проснулся");
        OnEnergyChanged($"{Name}: энергия {energyLevel}");
    }

    void IAnimal.Sleep()
    {
        energyLevel = 0;
        Console.WriteLine($"{Name} уснул");
        OnEnergyChanged($"{Name}: энергия {energyLevel}");
    }

    public override void WakeUp()
    {
        ((IAnimal)this).WakeUp();
        Console.WriteLine("Собака готова играть");
        OnActionPerformed($"{Name} проснулся и готов играть");
    }

    public override void Sleep()
    {
        ((IAnimal)this).Sleep();
        Console.WriteLine("Собака спит в будке");
    }

    public void Play()
    {
        energyLevel -= 20;
        Console.WriteLine($"{Name} играет с хозяином");
        OnEnergyChanged($"{Name}: энергия {energyLevel}");
    }

    public void Train()
    {
        if (!IsTrained)
        {
            IsTrained = true;
            energyLevel -= 10;
            Console.WriteLine($"{Name} обучен");
            OnActionPerformed($"{Name} прошёл обучение");
        }
        else
        {
            Console.WriteLine($"{Name} уже обучен");
        }
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}, Хозяин: {Owner}, Обучен: {IsTrained}");
    }
}

public class Cat : Animal, IPet
{
    public string Color { get; set; }
    public string Owner { get; set; }
    public bool IsIndoor { get; set; }

    public Cat(string name, int age, double price, string habitat, string color, string owner, bool isIndoor)
        : base(name, age, price, habitat)
    {
        Color = color;
        Owner = owner;
        IsIndoor = isIndoor;
    }

    public override void WakeUp()
    {
        energyLevel = 100;
        Console.WriteLine($"{Name} проснулся");
        OnEnergyChanged($"{Name}: энергия {energyLevel}");
        OnActionPerformed($"{Name} проснулся");
    }

    public override void Sleep()
    {
        energyLevel = 0;
        Console.WriteLine($"{Name} уснул");
        OnEnergyChanged($"{Name}: энергия {energyLevel}");
    }

    public void Play()
    {
        energyLevel -= 15;
        Console.WriteLine($"{Name} играет с хозяином");
        OnEnergyChanged($"{Name}: энергия {energyLevel}");
    }

    public void Purr()
    {
        energyLevel -= 5;
        Console.WriteLine($"{Name} мурлычет");
        OnActionPerformed($"{Name} мурлычет");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Окрас: {Color}, Хозяин: {Owner}, Домашняя: {IsIndoor}");
    }
}

public class Bird : Animal, IFlyable
{
    public string WingSpan { get; set; }
    public bool CanFly { get; set; }

    public Bird(string name, int age, double price, string habitat, string wingSpan, bool canFly)
        : base(name, age, price, habitat)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
    }

    public override void WakeUp()
    {
        energyLevel = 100;
        Console.WriteLine($"{Name} проснулся");
        OnEnergyChanged($"{Name}: энергия {energyLevel}");
        OnActionPerformed($"{Name} проснулся");
    }

    public override void Sleep()
    {
        energyLevel = 0;
        Console.WriteLine($"{Name} уснул");
        OnEnergyChanged($"{Name}: энергия {energyLevel}");
    }

    public virtual void Fly()
    {
        if (CanFly)
        {
            energyLevel -= 10;
            Console.WriteLine($"{Name} летает.");
            OnEnergyChanged($"{Name}: энергия {energyLevel}");
            OnActionPerformed($"{Name} летает");
        }
        else
        {
            Console.WriteLine($"{Name} не умеет летать.");
        }
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {WingSpan}, Умеет летать: {CanFly}");
    }
}

public class Parrot : Bird
{
    public string Vocabulary { get; set; }
    public int WordsCount { get; set; }

    public Parrot(string name, int age, double price, string habitat, string wingSpan, bool canFly,
                  string vocabulary, int wordsCount)
        : base(name, age, price, habitat, wingSpan, canFly)
    {
        Vocabulary = vocabulary;
        WordsCount = wordsCount;
    }

    public override void Fly()
    {
        energyLevel -= 5;
        Console.WriteLine($"{Name} летает по комнате.");
        OnEnergyChanged($"{Name}: энергия {energyLevel}");
        OnActionPerformed($"{Name} летает по комнате");
    }

    public void Speak()
    {
        Console.WriteLine($"{Name} говорит: {Vocabulary}");
        OnActionPerformed($"{Name} говорит: {Vocabulary}");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Словарный запас: {Vocabulary}, Количество слов: {WordsCount}");
    }
}

public class Penguin : Bird, ISwimmable
{
    public double SwimSpeed { get; set; }

    public Penguin(string name, int age, double price, string habitat, string wingSpan, bool canFly,
                   double swimSpeed)
        : base(name, age, price, habitat, wingSpan, canFly)
    {
        SwimSpeed = swimSpeed;
    }

    public override void Fly()
    {
        Console.WriteLine($"{Name} не умеет летать.");
    }

    public void Swim()
    {
        energyLevel -= 15;
        Console.WriteLine($"{Name} плавает cо скоростью {SwimSpeed:F2} км/ч.");
        OnEnergyChanged($"{Name}: энергия {energyLevel}");
        OnActionPerformed($"{Name} плавает");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Скорость плавания: {SwimSpeed:F2} км/ч");
    }
}

public class AnimalCollection<T> where T : Animal
{
    private HashSet<T> _items = new HashSet<T>();

    public bool Add(T animal)
    {
        return _items.Add(animal);
    }

    public void DisplayAll()
    {
        foreach (var animal in _items)
        {
            animal.DisplayInfo();
            Console.WriteLine();
        }
    }

    public void DisplayAllSortedByPrice()
    {
        foreach (var animal in _items.OrderBy(a => a.Price))
        {
            animal.DisplayInfo();
            Console.WriteLine();
        }
    }
}

Console.WriteLine("Делегаты");

Dog dogDelegate = new Dog("Лайка", 3, 25000.00, "Дом", "Овчарка", "Иван", true);
Cat catDelegate = new Cat("Мурка", 2, 15000.00, "Дом", "Белый", "Анна", true);

DisplayDelegate displayDog = dogDelegate.DisplayInfo;
DisplayDelegate displayCat = catDelegate.DisplayInfo;

displayDog();
displayCat();

Console.WriteLine("Лямбда-выражения");

List<Animal> animals = new List<Animal>
{
    new Dog("Лайка", 3, 25000.00, "Дом", "Овчарка", "Иван", true),
    new Dog("Отгадай", 2, 30000.00, "Дом", "Лабрадор", "Мария", false),
    new Dog("Котя", 4, 28000.00, "Дом", "Хаски", "Пётр", true),
    new Cat("Мурка", 2, 15000.00, "Дом", "Белый", "Анна", true),
    new Cat("Рыжик", 1, 12000.00, "Дом", "Рыжий", "Сергей", true),
    new Cat("Фелемонтий", 5, 18000.00, "Дом", "Серый", "Ольга", false),
    new Parrot("Кеша", 2, 20000.00, "Клетка", "25.5 см", true, "Привет, как дела?", 3),
    new Parrot("Гоша", 3, 22000.00, "Клетка", "28.0 см", true, "Кто там?", 2),
    new Penguin("Пингу", 4, 35000.00, "Антарктида", "40.0 см", false, 36.5),
};

var trainedDogs = animals.OfType<Dog>().Where(d => d.IsTrained);
foreach (var dog in trainedDogs)
{
    dog.DisplayInfo();
}

var expensiveAnimals = animals.Where(a => a.Price > 20000);
foreach (var animal in expensiveAnimals)
{
    Console.WriteLine($"{animal.Name}: {animal.Price:F2} Руб");
}

Console.WriteLine("===== События =====");

Dog eventDog = new Dog("Лайка", 3, 25000.00, "Дом", "Овчарка", "Иван", false);
eventDog.EnergyChanged += message => Console.WriteLine($"[Энергия] {message}");
eventDog.ActionPerformed += message => Console.WriteLine($"[Действие] {message}");

eventDog.WakeUp();
eventDog.Play();
eventDog.Train();
eventDog.Sleep();

Console.WriteLine("Коллекции");

AnimalCollection<Dog> dogCollection = new AnimalCollection<Dog>();
dogCollection.Add(new Dog("Лайка", 3, 25000.00, "Дом", "Овчарка", "Иван", true));
dogCollection.Add(new Dog("Отгадай", 2, 30000.00, "Дом", "Лабрадор", "Мария", false));
dogCollection.Add(new Dog("Котя", 4, 28000.00, "Дом", "Хаски", "Пётр", true));
dogCollection.Add(new Dog("Лайка", 3, 25000.00, "Дом", "Овчарка", "Иван", true));

Console.WriteLine("Коллекция собак");
dogCollection.DisplayAll();

Console.WriteLine("Коллекция собак по цене");
dogCollection.DisplayAllSortedByPrice();
```
